## Exercise 2: Character Classes, Anchors & Non-Matching Groups/

### 1. BeautifulSoup Extraction

```python
plain_text = soup.select_one("div.show-synopsis p").get_text(" ", strip= True)
print(plain_text)
```

###

#### Task 2A

## Exercise 3: Quantifiers (Greedy vs. Lazy) & Group Captures

### BeautifulSoup Extraction

```python
href = soup.find("a", string="Viki")["href"]
```

### Regex Extraction

#### Task 3A

```python
#<a href="/redirect?q=https%3A%2F%2Fwww.viki.com%2Ftv%2F32311c&utm_source=mdl"rel="external nofollow"><b>Viki</b></a>

domain = r"https%3A%2F%2Fwww\.(.+?)%2F"
print(re.search(domain,href).group(1))
```

#### Task 3B

`re.search()` is used for creating a match object, and `.group()` or `.groups()` is used for retrieving captured groups

```python
three_group = r"(https)%3A%2F%2Fwww\.(.+?)%2Ftv%2F(\w+)"
print(re.search(three_group,href).groups())
```

## Exercise 4: Lookaheads & Lookbehinds (Assertions)

### BeautifulSoup Extraction

```python
container = soup.select_one("div.col-xs-8.col-sm-9.col-lg-9").get_text(" ",strip=True)
print(container)
```

### Regex Extraction

```python
pattern = (
    r"Ratings:\s*(?P<rating>\d+\.\d+)"
r"\s*/10 from\s*(?P<rating_users>[\d,]+)\s*users"
r".*?# of Watchers:\s*(?P<watchers>[\d,]+)"
r".*?Reviews:\s*(?P<review_users>[\d,]+)\s*users")

print(re.search(pattern, container).groupdict())
```
