```python
from bs4 import BeautifulSoup
import re

with open('bongsoon.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

Check that the parsing was successful:

```python
soup.title
```

## Exercise 1

### 1. BeautifulSoup Extraction

```python
content = soup.find("meta",attrs={"name":"description"})["content"]
print(content)
```

### Regex Extraction

###

#### Task 1A

```python
release_year = r"\b\d{4}\b"
```

#### Task 1B

```python
six_letter = r"\b\w{6}\b"
```

#### Task 1C

```python
print(re.findall(release_year,content))
```

```python
print(re.findall(six_letter,content))
```

## Exercise 2: Character Classes, Anchors & Non-Matching Groups/

### 1. BeautifulSoup Extraction

```python
plain_text = soup.select_one("div.show-synopsis p").get_text(" ", strip= True)
print(plain_text)
```

### Regex Extraction

###

#### Task 2A

```python
vowel_start = r"\b[aeiou]\w+\b"
print(re.findall(vowel_start,plain_text))
```

#### Task 2B

```python
no_ae = r"\b[^ae\W]+\b"
print(re.findall(no_ae,plain_text))
```
